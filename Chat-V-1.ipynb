### Title

- Rule-Based AI Chatbot using Python

### Objective

- To create a chatbot that can respond to basic user queries.

### Concept

- Uses if-else conditions (rule-based AI)

- Matches user input with predefined responses

```python
def chatbot():
    print("Chatbot: Hello! Type 'bye' to exit.")

    while True:
        user = input("You: ").lower().strip()

        if not user:
            continue

        if "hello" in user:
            print("Chatbot: Hi! How can I help you?")
        
        elif "how are you" in user:
            print("Chatbot: I am fine!")
        
        elif "what is ai" in user:
            print("Chatbot: AI means Artificial Intelligence.")
        
        elif "bye" in user:
            print("Chatbot: Goodbye!")
            break
        
        else:
            print("Chatbot: Sorry, I don't understand.")

chatbot()
```

#### Steps

1. Take user input

2. Match with conditions

3. Print response

4. Repeat until exit

#### Conclusion

- This chatbot demonstrates a basic AI interaction system using simple logic.
